# Continuous Linear Optimal Transport Transform 3D

This tutorial will demonstrate: how to use the $multVOT()$ to calculate CLOT for 3D images.

## Functions:
  
1. Forward transform:

    result = multVOT(I0, I1)

        Input:
        ----------------
        I1: image to transform
        I0: reference of transport maps

        Output:
        ----------------
        result: a dict
        --------
        result['I1'] : image to transform
        result['I0'] : reference of transport maps
        result['f1'] : transport maps in the x-direction
        result['f2'] : transport maps in the y-direction
        result['f3'] : transport maps in the z-direction



## Definition
The CLOT is solving the following problem:

$$f^*_i(\textbf{x}) = \arg \min_{f_i} \int |\textbf{x}-f_i(\textbf{x})|^2I_0(\textbf{x})d\textbf{x},$$

$$ s.t.\ det(Df_i(\textbf{x}))
    I_i(f_i(\textbf{x})) = I_0(\textbf{x})$$

Here, $D$ denotes the Jacobian matrix, and the $MP$ is the family of all mass-preserving mappings from $I_0$ to $I_i$. Here, $f_i : Ω ⟶ Ω$ is a mass preserving mapping from $I_0$ to $I_i$. The OT mapping is calculated from the reference image $I_0$.

The transport map $f_i^*(x)$ is the embeding of $I_i$ in the transport space.

```python
import numpy as np

I0 = np.load('data/CLOT3D/CLOT3D_I0.npy')
I1 = np.load('data/CLOT3D/CLOT3D_I1.npy')

# normalization
I0 = I0/np.sum(I0) + 1e-5
I1 = I1/np.sum(I1) + 1e-5
```

```python
from pytranskit.TBM3D.multVOT import multVOT

result = multVOT(I0, I1)
```

```python
from pytranskit.TBM3D.visualization import visualize_result
visualize_result(result)
```
